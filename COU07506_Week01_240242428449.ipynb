## 5. Background Theory

A discrete-time signal, represented as x[n], is defined only at integer values of n.

In digital signal processing, a discrete-time signal can be stored using two arrays:

1. The sample index, n.
2. The signal values, x[n].

A sampled audio file is represented by an array of numerical samples together with its sampling frequency, fs.

The sampling frequency indicates the number of samples taken per second and is measured in Hertz (Hz).


### 8.3 Generation of Elementary Discrete-Time Signals

In this experiment, four elementary discrete-time signals are generated using NumPy.

The sample index ranges from -5 to 10. Each signal is calculated using its mathematical definition and then visualized using a stem plot.


```python
# Generate sample indices
n = np.arange(-5, 11)

# Generate elementary signals
delta = (n == 0).astype(float)
step = (n >= 0).astype(float)
ramp = n * step
expo = 0.8**n * step

# Plot the four signals
fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))

for ax, x, title in zip(
    axes,
    [delta, step, ramp, expo],
    ["Unit Impulse", "Unit Step", "Unit Ramp", "Exponential"]
):
    ax.stem(n, x, basefmt="k-")
    ax.set_title(title)
    ax.set_xlabel("Sample index (n)")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```


### 8.4 Results and Discussion

The four elementary discrete-time signals were successfully generated and plotted using Python.

The following observations were made:

1. The unit impulse signal has a value of one at n = 0 and zero at all other sample indices.
2. The unit step signal changes from zero to one at n = 0 and remains at one for positive sample indices.
3. The unit ramp signal increases linearly for positive sample indices.
4. The exponential signal decreases gradually because its exponential parameter is 0.8, which is between zero and one.

The stem plots clearly represent the individual samples of each discrete-time signal.

### Conclusion

The experiment demonstrated how elementary discrete-time signals can be generated and visualized using NumPy and Matplotlib. It also provided an understanding of the different behaviours of impulse, step, ramp and exponential signals.



### Task 1.2 — Change the Parameters

In this experiment, the exponential signal x[n] = a^n u[n] is generated using three different values of the parameter a: 0.5, -0.8 and 1.1.

The resulting signals are plotted using subplots to compare their behaviour.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample indices
n = np.arange(-5, 11)

# Generate unit step
u = (n >= 0).astype(float)

# Generate exponential signals
x1 = (0.5**n) * u
x2 = ((-0.8)**n) * u
x3 = (1.1**n) * u

# Plot the signals
fig, axes = plt.subplots(1, 3, figsize=(13, 3.5))

axes[0].stem(n, x1)
axes[0].set_title("a = 0.5")

axes[1].stem(n, x2)
axes[1].set_title("a = -0.8")

axes[2].stem(n, x3)
axes[2].set_title("a = 1.1")

for ax in axes:
    ax.set_xlabel("Sample index (n)")
    ax.set_ylabel("Amplitude")
    ax.grid(True)

plt.tight_layout()
plt.show()
```


### 9. Results and Discussion

Three exponential signals were generated using different values of the parameter a.

The following observations were obtained:

1. When a = 0.5, the signal decreases gradually towards zero. This is called a decaying exponential signal.

2. When a = -0.8, the signal alternates between positive and negative values while its magnitude decreases gradually. This is called an alternating decaying exponential signal.

3. When a = 1.1, the signal increases as the sample index increases. This is called a growing exponential signal.

The results demonstrate that the value and sign of the exponential parameter determine the behaviour of a discrete-time exponential signal.

### Conclusion

The experiment demonstrated how changing the exponential parameter affects a discrete-time signal. Positive values of a between zero and one produce decaying signals, negative values between minus one and zero produce alternating decaying signals, while values greater than one produce growing signals.



### Task 1.2 (Part 2) — Cosine Signals and Periodicity

Two discrete-time cosine signals are generated using different periods in their mathematical expressions.

The signals are plotted for sample indices from 0 to 40 to investigate whether their values repeat.


```python

import numpy as np
import matplotlib.pyplot as plt

# Generate sample indices
n = np.arange(0, 41)

# Generate cosine signals
x1 = np.cos(2 * np.pi * n / 8)
x2 = np.cos(2 * np.pi * n / 7.5)

# Plot the first cosine signal
plt.figure(figsize=(10, 4))
plt.stem(n, x1)
plt.title("Cosine Signal 1: Fundamental Period = 8")
plt.xlabel("Sample index (n)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()

# Plot the second cosine signal
plt.figure(figsize=(10, 4))
plt.stem(n, x2)
plt.title("Cosine Signal 2: Fundamental Period = 15")
plt.xlabel("Sample index (n)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```


### Results and Discussion

Two discrete-time cosine signals were generated and plotted for sample indices from 0 to 40.

The observations were:

1. The first cosine signal, cos(2πn/8), repeats after every 8 samples. Its fundamental period is 8.

2. The second cosine signal, cos(2πn/7.5), repeats after every 15 samples. Its fundamental period is 15.

Both signals are periodic because each has a positive integer fundamental period.

### Conclusion

The experiment demonstrated that discrete-time cosine signals can have different fundamental periods depending on their angular frequencies. It also showed that the period of a discrete-time signal must be a positive integer number of samples.



### Task 1.3 — Working with Audio

A voice recording was captured using a smartphone and converted from M4A to WAV format.

The WAV file was uploaded to Google Colab and read using the SciPy library.

The audio signal was converted to mono where necessary, normalized, plotted against time and played back using the IPython Audio module.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io import wavfile
from IPython.display import Audio, display

# Read the WAV file
fs, x = wavfile.read("Recording.wav")

# Convert stereo audio to mono if necessary
if x.ndim > 1:
    x = x.mean(axis=1)

# Convert samples to floating-point values
x = x.astype(float)

# Normalize the audio safely
peak = np.max(np.abs(x))

if peak > 0:
    x = x / peak

# Generate the time axis
t = np.arange(len(x)) / fs

# Display recording information
print("Sampling frequency:", fs, "Hz")
print("Number of samples:", len(x))
print("Duration:", len(x)/fs, "seconds")

# Plot the audio waveform
plt.figure(figsize=(12, 4))
plt.plot(t, x)
plt.title("My Voice Recording")
plt.xlabel("Time (seconds)")
plt.ylabel("Normalized Amplitude")
plt.grid(True)
plt.show()

# Play the recording
display(Audio(x, rate=fs))
```


### Results and Discussion

The recorded audio file was successfully loaded and processed using Python.

The following results were obtained:

- Sampling frequency: 48,000 Hz
- Number of samples: 179,200
- Recording duration: 3.7333 seconds
- Audio channels: Mono

The waveform showed variations in amplitude over time, representing the recorded voice.

The signal was normalized to approximately the range -1 to +1 and successfully played back in Google Colab.

### Duration Calculation

The recording duration is calculated using:

T = N / fs

T = 179200 / 48000

T = 3.7333 seconds

### Conclusion

The experiment demonstrated how digital audio can be loaded, normalized, visualized and played back using Python. It also introduced the relationship between sampling frequency, number of samples and recording duration.
